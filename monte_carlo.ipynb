```python
import numpy as np
```

```python
N = 1000000
U_samples = np.random.uniform(0, 1, N)
```

```python
np.mean(np.sin(np.log(1+U_samples**17)))
```

```python
# integrate sin(cos(sin(x))) from 0 to 3, monte-carlo method
N = 1000000
U_samples = np.random.uniform(0, 3, N)
np.mean(np.sin(np.cos(np.sin(U_samples))))
```

```python
# integrate log(1+x**2)*exp(-x**2/2) from -inf to inf
N = 1000000
U_samples = np.random.normal(0, 1, N)
np.mean(np.log(1+U_samples**2)*np.exp(-U_samples**2/2))
```

```python
# find p(0.1<X**2<0.2), X ~ N(0,1), monte-carlo method
N = 1000000
U_samples = np.random.normal(0, 1, N)
np.mean((U_samples**2 > 0.1) & (U_samples**2 < 0.2))
```

```python
# find E(sin(x**2)), X ~ N(0,1), monte-carlo method
N = 1000000
U_samples = np.random.normal(0, 1, N)   
np.mean(np.sin(U_samples**2))
```
