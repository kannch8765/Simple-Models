```python
import numpy as np
import pylab as plt
import jax
```

```python
n=500
d=300
X = np.random.normal(loc=0, scale=1., size=(n,d))
# generate a random n*d matrix, entries follow normal distribution
# with mean 0 and sd 1
X.shape
```

```python
cov_X=np.cov(X.T)
cov_X.shape
```

```python
plt.imshow(cov_X)
plt.colorbar()
```

```python
eigenvalues, eigenvectors = np.linalg.eigh(cov_X) 
#eigenvalues. vectors for real symmetric matrix
```

```python
plt.hist(eigenvalues)
```

```python
A = np.random.normal(loc=0, scale=1, size=(2000,2000))
M= 0.5 * (A+A.T)
eigenvalues_M, _ = np.linalg.eigh(M)
plt.hist(eigenvalues_M, bins=100);
```

```python
import time
```

```python
d = 50
R = np.random.normal(0,1,size=(d,d))
M = np.identity(d)+ R @ R.T

start = time.time()
M_inv=np.linalg.inv(M)
end = time.time()

print(f"Time taken: {end - start}")
```

```python
dim_list = [50, 100, 200, 500, 700, 1000, 2000, 3000, 5000, 7000]
time_list = []

for d in dim_list:
    print(f"Dimension: {d}")
    R = np.random.normal(0,1,size=(d,d))
    M = np.identity(d)+ R @ R.T

    start = time.time()
    M_inv=np.linalg.inv(M)
    end = time.time()

    time_list.append(end-start)
    print(f"\t It took {end-start: 2.4f} sec")
```

```python
plt.plot(dim_list,time_list,"-^")
plt.xscale("log")
plt.yscale("log")
plt.grid()
```
