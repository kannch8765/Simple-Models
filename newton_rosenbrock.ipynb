```python
import numpy as np
import matplotlib.pyplot as plt
from jax.config import config
config.update("jax_enable_x64", True)
import jax
import jax.numpy as jnp
```

```python
#rosenbrock function
def rosenbrock(x):
    return 100*(x[1]-x[0]**2)+ (1-x[0])**2
```

```python
grad_r = jax.grad(rosenbrock)
hess_r = jax.hessian(rosenbrock)

def grad_r_numpy(x):
    return np.array(grad_r(x))

def grad_num(grad):
    return np.array(grad)
```

```python
import scipy.optimize as opt
```

```python
# use l-bfgs-b method
x_init = np.random.normal(0,1,size=2   )
opt.minimize(rosenbrock,
             x_init,
             jac=grad_r,
             method='L-BFGS-B',)
```
